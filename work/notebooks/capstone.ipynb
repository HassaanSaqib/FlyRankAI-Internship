## 1. Question

*The research question and the decision it supports.*

Can a machine learning model help rank content pages that should be reviewed for potential refresh opportunities using anonymized search-performance signals?

The goal is to support content-review decisions by identifying pages with higher observed decline risk. The output is intended as decision support for human reviewers rather than an automated content-management system

## 2. Data

*Which release, which tables, date windows, what you excluded and why. Public-safe.*

This project uses the FlyRank ML Internship anonymized starter dataset.

The analysis uses anonymized page-level search-performance measurements together with retrospective decline labels. Direct identifiers, client names, URLs, private queries and credentials were excluded throughout the project.

The work focuses on public-safe measurements and grouped validation to reduce information leakage.


## 3. Methodology

*Assumptions, features, label definition, baseline, validation design, leakage checks.*

The project follows the Refresh / Content Opportunity Scoring lane.

The workflow consisted of:

- defining the prediction task
- creating a rule-based baseline
- training Logistic Regression and Random Forest models
- validating performance using grouped client holdout
- auditing feature leakage
- comparing the model against the Week-4 baseline
- generating a ranked human-review playbook

Strict feature selection and leakage checks were applied before producing the final recommendations.

## 4. Results (vs baseline)

*Model vs baseline on the same split. The honest table.*

The trained machine-learning models were compared with the Week-4 rule-based baseline using the same grouped validation split and evaluation metrics.

Performance was assessed using Precision@20, Precision@50, Average Precision and ROC AUC.

The selected model produced the strongest overall ranking performance and formed the basis of the final recommendation system. All reported results are observational and are intended for decision support only.


## 5. Limitations

*What this work cannot claim.*

This project does not claim to model Google's ranking algorithm or prove that refreshing content causes future performance improvements.

The work is limited to the anonymized internship dataset and retrospective labels. Recommendations require human review before any editorial or technical action is taken.

## 6. Ranked recommendations

*The action playbook output — the paper's recommendations section.*

The final output is a ranked review queue supported by transparent reason codes.

Recommended actions include:

- Refresh Review
- Metadata Review
- Diagnose First
- Monitor and Review
- Protect and Monitor
- Data Quality Review

Each recommendation includes human-readable reasoning and requires manual review before implementation.

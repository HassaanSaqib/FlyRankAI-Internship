## Quick Look at the Data

I will use the starter dataset to understand whether this research lane is worth exploring.

The code below loads the dataset and reports a few key statistics. These numbers will help me understand the scale of the data and whether there are enough examples to investigate this problem further.

### Findings

The starter dataset contains **200 pages**. All **200 pages** have recorded impressions, and **197 pages (98.5%)** are marked with a downward trend.

These numbers suggest there are enough examples to investigate this research question. A system that helps prioritize which pages should be reviewed first could support better decision-making, although these observations do not prove that updating a page will necessarily improve its performance.

```python
# This cell is for CODE (numbers, a query, a check).
# Write your text answer in the cell ABOVE this one — typing sentences here breaks Run All.
import os

import glob

import subprocess

import pandas as pd

repo_dir = "/content/flyrank_repo"

# Download your repository into Colab so the dataset is available

if not os.path.exists(repo_dir):

    subprocess.run(

        [

            "git",

            "clone",

            "-q",

            "https://github.com/HassaanSaqib/FlyRankAI-Internship.git",

            repo_dir,

        ],

        check=True,

    )

# Search the repository for a CSV containing the required starter columns

csv_files = glob.glob(f"{repo_dir}/**/*.csv", recursive=True)

df = None

data_path = None

for path in csv_files:

    try:

        test_df = pd.read_csv(path)

        required_columns = {"impressions_90d", "trend_direction"}

        if required_columns.issubset(test_df.columns):

            df = test_df

            data_path = path

            break

    except Exception:

        continue

if df is None:

    raise FileNotFoundError(

        "No starter CSV with impressions_90d and trend_direction was found."

    )

total_pages = len(df)

down_pages = int(df["trend_direction"].astype(str).str.lower().eq("down").sum())

down_percent = (down_pages / total_pages) * 100

pages_with_impressions = int((df["impressions_90d"] > 0).sum())

print("Dataset loaded successfully.")

print(f"Total pages: {total_pages}")

print(f"Pages with impressions above 0: {pages_with_impressions}")

print(f"Pages with a downward trend: {down_pages} ({down_percent:.1f}%)")
```
